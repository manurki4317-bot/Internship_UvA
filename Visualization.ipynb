Usa `hoi.plots`. Para el visor 3D interactivo, abre `viz/hoi_hub_viewer.html` o regéneralo con `viz/build_viewer.py`.

```python
import sys
sys.path.insert(0, '..')
import numpy as np, pandas as pd
import hoi
from hoi import atlas as A, plots
%matplotlib inline
```

```python
# >>> AJUSTA ESTA RUTA a la carpeta donde tengas los CSV (por defecto: Average_Data/ en la raíz del paquete)
DATA = '../Average_Data'

df   = pd.read_csv(f'{DATA}/average_triplets.csv', index_col=0)
null = pd.read_csv(f'{DATA}/aggregated_centralities1000copies0005fraction.csv')
atlas, brain = A.load_atlas(), A.load_brain()
```

```python
top  = hoi.select_triplets(df, column='Mut Info_normalized', mode='redundancy')   # 'synergy' para sinergia
cent = hoi.centralities(hoi.adjacency(top['nplets']), top['nplets'])
thr  = hoi.hub_thresholds(null, p=0.05)
is_hub = hoi.exceeds(cent['EC_z'], thr['EC_z'])
tri = np.array(top['nplets'].tolist()) - 1     # los CSV usan índices 1-based
is_hub.sum()
```

```python
plots.plot_hub_projections(atlas, tri, is_hub, brain);
```

```python
plots.plot_pair_matrix(atlas, tri[is_hub]);
```

```python
plots.plot_null_vs_real(null['EC_z'], cent['EC_z'], thr['EC_z']);
```

```python
plots.plot_pairwise_vs_highorder(atlas, [26, 60, 21, 65, 22, 64, 24, 25, 61, 63], tri[is_hub]);   # hubs por pares: CodeBlock5
```
